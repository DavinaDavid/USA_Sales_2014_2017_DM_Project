```python
import pandas as pd
import os
```

```python
project_root = os.getcwd()
processed_folder = os.path.join(project_root, "data", "processed")
# Read retail Excel data
order_sales_on_holiday_file = os.path.join(processed_folder, "retail_with_holidays.csv")
order_sales_on_holidays = pd.read_csv(order_sales_on_holiday_file)
```

```python
#=====================================================
#STEP 8: DATA PROFILING / QUALITY CHECK
#=====================================================
print("\nData Quality Checks:")

print("Missing Sales:",
      order_sales_on_holidays["Sales"].isna().sum())

print("Missing Is_Holiday:",
      order_sales_on_holidays["Is_Holiday"].isna().sum())

print("Duplicate rows:",
      order_sales_on_holidays.duplicated().sum())

print("\nSales Statistics:")
print(order_sales_on_holidays["Sales"].describe())
```
