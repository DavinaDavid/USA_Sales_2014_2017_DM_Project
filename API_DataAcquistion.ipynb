```python
import json
import requests
import pandas as pd
```

```python
# -----------------------------
# STEP 3: Fetch US holidays 2011 via API
```

```python
years = [2014, 2015, 2016, 2017]
api_urls = [f"https://date.nager.at/api/v3/PublicHolidays/{year}/us" for year in years]
```

```python
holiday_data_list = []

for url in api_urls:
    response = requests.get(url)
    us_holidays_raw = response.json()  # parse JSON into Python list of dicts
    us_holidays_raw_df = pd.DataFrame(us_holidays_raw)
    holiday_data_list.append(us_holidays_raw_df)
```

```python
# Combine all years into one DataFrame
merged_holidays = pd.concat(holiday_data_list, ignore_index=True)

# Optional: keep only columns we need
merged_holidays = merged_holidays[['date', 'localName', 'name']]

# Convert 'date' to datetime
merged_holidays['date'] = pd.to_datetime(merged_holidays['date'])
```

```python
# Convert DataFrame to list of dictionaries
merged_holidays_dict = merged_holidays.to_dict(orient="records")
```

```python
import os
project_root = os.getcwd()
raw_folder = os.path.join(project_root, "data", "raw")
```

```python
# Make sure 'date' column is string in ISO format
merged_holidays_dict = merged_holidays.copy()
merged_holidays_dict['date'] = merged_holidays_dict['date'].dt.strftime('%Y-%m-%d')
```

```python
# Convert to list of dicts
merged_holidays_list = merged_holidays_dict.to_dict(orient="records")

project_root = os.getcwd()
raw_folder = os.path.join(project_root, "data", "raw")

# Output file path save to JSON
output_file = os.path.join(raw_folder, "merged_holidays_api.json")

# Dump JSON
with open(output_file, "w") as f:
    json.dump(merged_holidays_list, f, indent=4)

print(f"Merged holidays saved to {output_file}")
```

```python
merged_holidays_df = pd.DataFrame(merged_holidays_list)
output_csv = os.path.join(raw_folder, "merged_holidays_api.csv")

# Save to CSV
merged_holidays_df.to_csv(output_csv, index=False)

print(f"Merged holidays saved as CSV to {output_csv}")
```
