```python
import os
import pandas as pd
```

```python
project_root = os.getcwd()
processed_folder = os.path.join(project_root, "data", "processed")
# Read retail Excel data
retail_file = os.path.join(processed_folder, "cleaned_retail_data.csv")
retail_data = pd.read_csv(retail_file)

# Read holidays Excel/API data
holidays_file = os.path.join(processed_folder, "cleaned_holidays.csv")
holidays = pd.read_csv(holidays_file)
```

```python
print("Files loaded successfully!")
print("Retail shape:", retail_data.shape)
print("Holidays shape (before cleaning):", holidays.shape)
```

```python
# STEP 2: Ensure correct date formats
# -----------------------------

# Convert to datetime
retail_data['Order Date'] = pd.to_datetime(retail_data['Order Date'])
holidays['date'] = pd.to_datetime(holidays['date'])

# Keep only date part (remove time component)
retail_data['Order Date'] = retail_data['Order Date'].dt.date
holidays['date'] = holidays['date'].dt.date
```

```python
# -----------------------------
# STEP 3: Check matching dates (sanity check)
# -----------------------------

common_dates = set(retail_data['Order Date']) & set(holidays['date'])
print(f"Number of matching dates: {len(common_dates)}")
print("Some matching dates:", list(common_dates)[:10])
```

```python
# -----------------------------
# STEP 4: Create Holiday Flag in Retail Data (IMPORTANT)
# -----------------------------

retail_data['Is_Holiday'] = retail_data['Order Date'].isin(holidays['date'])
retail_data['Is_Holiday'] = retail_data['Is_Holiday'].astype(int)

print("\nSample check of holiday flag:")
print(retail_data[['Order Date', 'Is_Holiday']].head(10))
```

```python
# -----------------------------
# STEP 5: Prepare Holidays dataset for merging
# -----------------------------

# REMOVE DUPLICATES → VERY IMPORTANT (prevents row explosion)
holidays = holidays.drop_duplicates(subset=['date'])

print("Holidays shape after removing duplicates:", holidays.shape)
```

```python
# -----------------------------
# STEP 6: Merge (Data Integration Step)
# -----------------------------

order_purchases_on_holidays = retail_data.merge(
    holidays,
    left_on='Order Date',
    right_on='date',
    how='left'
)

print("\nShape after merge:", order_purchases_on_holidays.shape)
```
