```python
import numpy as np
import pandas as pd
import sqlite3
import os
```

```python
project_root = os.getcwd()
processed_folder = os.path.join(project_root, "data", "processed")
# Read retail Excel data
order_purchases_on_holiday_file = os.path.join(processed_folder, "retail_with_holidays.csv")
order_purchases_on_holidays = pd.read_csv(order_purchases_on_holiday_file)
```

```python
#-----------------------------------------------------
#STEP 6: Make DataFrames SQLite-compatible
#-----------------------------------------------------
def make_sqlite_compatible(df):
    for column in df.columns:
        if df[column].apply(
            lambda x: isinstance(x, (list, dict, np.ndarray))
        ).any():
            df[column] = df[column].apply(
                lambda x: str(x) if pd.notnull(x) else x
            )
    return df
```

```python
# Apply to your integrated dataset
order_sales_on_holidays = make_sqlite_compatible(order_sales_on_holidays)

print("SQLite compatibility ensured")
```

```python
#-----------------------------------------------------
#STEP 7: Store integrated data in SQLite DB
#-----------------------------------------------------
db_folder = os.path.join(project_root, "db")
db_path = os.path.join(db_folder, "usa_retail_holidays_2014_2017.db")
conn = sqlite3.connect(db_path)

order_purchases_on_holidays.to_sql(
    "retail_sales_holidays",
    conn,
    if_exists="replace",
    index=False
)

print("Data stored in SQLite DB at:", db_path)
```
